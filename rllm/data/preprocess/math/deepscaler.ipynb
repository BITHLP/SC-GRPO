```python
import json

datasets = ["amc", "omni_math", "aime", "still", "math"]

data = []
for dataset in datasets:
    with open(f"../train/{dataset}.json") as f:
        data.extend(json.load(f))
# with open(f'../train/omni_math.json', 'r') as f:
#     data.extend(json.load(f))
len(data)
```

```python
from rllm.rewards.math_utils.utils import _normalize, should_allow_eval

filtered_data = []
count = 0
for entry in data:
    answer = entry["answer"]
    try:
        if should_allow_eval(_normalize(answer)):
            filtered_data.append(entry)
            continue
        else:
            # print(answer)
            count += 1
    except Exception:
        # print(answer)
        count += 1

count
```

```python
with open("./deepscaler.json", "w") as f:
    json.dump(filtered_data, f)
```
